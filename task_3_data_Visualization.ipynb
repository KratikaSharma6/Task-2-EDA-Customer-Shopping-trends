# Project Overview
This project focuses on presenting customer shopping data using different types of visualization .
the main purpse is to make the information easier to understand and identify useful patterns from the dataset.
In this task,I will use charts and graphs to visualize customer demographics , product categories, purchase amounts,ratings,payment methods and shopping frequency.
## Objective
The objective of this task is to create clear and meaningful visualization that help in understanding customer shopping behaviour and comparing different aspects of the dataset.    

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("/content/shopping_trends.csv")
df.head()
```

```python
plt.figure(figsize=(8,5))
sns.countplot(data=df , x="Category")
plt.title("Category Distribution")
plt.xlabel("Category")
plt.ylabel("Count")
plt.show()
```

### Observation

From the chart, I can see that Clothing has the highest number of purchases in the dataset. Accessories is the next most common category, while Footwear and Outerwear have fewer records.

This shows that the number of purchases is not the same across all product categories.

```python
category_purchase = df.groupby("Category")["Purchase Amount (USD)"].mean()
print(category_purchase)
```

```python
plt.figure(figsize=(8,5))

category_purchase.plot(kind="bar")

plt.title("Average Purchase Amount by Category")
plt.xlabel("Product Category")
plt.ylabel("Average Purchase Amount (USD)")
plt.xticks(rotation=0)

plt.show()
```

### Observation

The average purchase amount is quite similar across all four categories. Footwear has the highest average purchase amount, while Outerwear has the lowest. The difference between the categories is not very large.

```python
payment_counts = df["Payment Method"].value_counts()

print(payment_counts)
```

```python
plt.figure(figsize=(9,5))
payment_counts.plot(kind="bar")
plt.title("Payment Method Distribution")
plt.xlabel("Payment Method")
plt.ylabel("number of customers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

### OBSERVATION
From the chart, I can see that credit card is the most commonly used payment method in the dataset.
Vemo and cash are also used by many customers, while the other payment methods have slightly fewer customers.
Overall, the difference between the payment methods is not very large.       

```python
frequency_counts = df["Frequency of Purchases"].value_counts()

print(frequency_counts)
```

```python
plt.figure(figsize=(9,5))
frequency_counts.plot(kind="bar")
plt.title("Shopping Frequency of Customers")
plt.xlabel("Frequency of purchase")
plt.ylabel("Number of Customers")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
category_rating = df.groupby("Category")["Review Rating"].mean().sort_values(ascending=False)
plt.figure(figsize =(8,5))
category_rating.plot(kind="bar")
plt.title("Average Review Rating by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Average Review Rating")
plt.xticks(rotation=0)
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8,5))

sns.scatterplot(data=df , x="Age" , y="Purchase Amount (USD)",alpha=0.5)
plt.title("Age vs Purchase Amount")
plt.xlabel("Age")
plt.ylabel("Purchase Amount (USD)")
plt.tight_layout()
plt.show()
```

### OBSERVATION
The points are spread across different ages and purchase amounts without showing a clear pattern.This suggests that age does not have a strong visible relationship with purchase amount in this dataset.

```python
numeric_columns =[
    "Age",
    "Purchase Amount (USD)",
    "Review Rating",
    "Previous Purchases"
]

Correlation_matrix = df[numeric_columns].corr()
plt.figure(figsize=(8,6))
sns.heatmap(Correlation_matrix,annot=True,cmap="coolwarm",fmt=".2f")
plt.title("Correlation between numerical variables")
plt.tight_layout()
plt.show()
```
